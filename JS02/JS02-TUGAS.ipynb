```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
dpath = '/content/wbc.csv'
df = pd.read_csv(dpath)
df.head()
```

```python
df.info()
```

```python
df.columns
```

1. Memisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan

```python
unused_columns = ['id', 'Unnamed: 32']
df = df.drop(columns=unused_columns)
df.head()
```

```python
df.columns
```

2. Melakukan proses encoding pada kolom "diagnosis".

```python
print("Keterangan Diagnosis:")
print("M = Malignant")
print("B = Benign")

print("\nNilai unik pada diagnosis:")
print(df['diagnosis'].unique())
```

```python
contoh_diagnosis = df.groupby('diagnosis').head(3)

contoh_diagnosis[['diagnosis']]
```

```python
diagnosis_awal = df['diagnosis'].copy()
```

```python
le = LabelEncoder()
df['diagnosis'] = le.fit_transform(df['diagnosis'])
```

```python
print("Mapping Encoding:")

for label, angka in zip(le.classes_, le.transform(le.classes_)):
    print(f"{label} = {angka}")
```

```python
hasil_encoding = pd.DataFrame({
    'Diagnosis Awal': diagnosis_awal,
    'Arti Diagnosis': diagnosis_awal.map({
        'M': 'Malignant',
        'B': 'Benign'
    }),
    'Hasil Encoding': df['diagnosis']
})

hasil_encoding.groupby('Diagnosis Awal').head(3)
```

3. Melakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.

```python
X = df.drop(columns=['diagnosis'])
y = df['diagnosis']
```

```python
X.head()
```

```python
y.head()
```

```python
X.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

X_scaled.head()
```

```python
df_scaled = X_scaled.copy()
df_scaled['diagnosis'] = y.values
df_scaled.head()
```

```python
df_scaled.info()
```

```python
df_scaled.head()
```

```python
print("Rata-rata setelah standardisasi:")
print(X_scaled.mean().head())

print("\nStandar deviasi setelah standardisasi:")
print(X_scaled.std().head())
```
